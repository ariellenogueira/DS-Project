# Part 1: Loading Data

```python
# grain size data
## this is a large file but you only need two of the columns
HOLY = np.genfromtxt("Data_HOLY4.csv", delimiter=",", skip_header=1)
GS_depth = HOLY[:,7] # depth in core
coarse_raw = HOLY[:,8] # raw coarse %

# age data
## this comes from running the age model in R
ages = np.genfromtxt("Ages_HOLY4.txt", skip_header=1)
ages_depth = ages[:, 0]

# interpolate ages onto GS_depth 
## in GS_depth, the depths are recorded as middle depths (0.5 cm, 1.5 cm,...) 
## while in the age document, they're recorded as top depths (0 cm, 1 cm,...)
ages_min  = np.interp(GS_depth, ages_depth, ages[:, 1])
ages_max  = np.interp(GS_depth, ages_depth, ages[:, 2])
ages_med  = np.interp(GS_depth, ages_depth, ages[:, 3])
ages_mean = np.interp(GS_depth, ages_depth, ages[:, 4])

# making sure this works, it should print as 853 853
print(len(GS_depth), len(ages_med))
```

## 1.2 Making a Time Series

### 1.2.1 Get Sliding Window

```python
# Considering grain size anomaly instead of raw grain size helps to account for changes in the background sediment size.

## From our age model, we find that we have a 6-10 yr/cm accumulation rate, so we can use an 8 yr/cm average.
## Assuming this, we can apply a 13 cm sliding window, which will give a centennial average.
window_size = 13 

# Convert array to pandas series for rolling window.
coarse_series = pd.Series(coarse_raw)
  
# Get the windows and moving average.
windows = coarse_series.rolling(window_size)
mov = windows.mean()

# Fill the NaNs at the start of the moving average with the series mean.
avg_coarse = np.mean(coarse_raw)
moving_averages = mov.fillna(avg_coarse)
  
# Convert pandas series back to numpy.
coarse_mov = moving_averages.to_numpy()
```

### 1.2.2 Get GS Anomaly

```python
anomaly_coarse = coarse_raw - coarse_mov

# Plot this to make sure it worked.
fig, ax = plt.subplots(2, 1, figsize=(12, 6), dpi=150, constrained_layout=True)
ax0 = ax[0]
ax1 = ax[1]

# Coarse % with moving average
ax0.plot(GS_depth,coarse_raw, label='HOLY4 Coarse %')
ax0.plot(GS_depth, coarse_mov, label=f'{window_size}pt mov avg')
ax0.grid(color='lightgray', linestyle=':', linewidth=0.7, axis='both', which='major', zorder=0)
ax0.set_ylabel('Coarse (%)')
ax0.set_xlabel('depth (cm)')
ax0.set_title('HOLY4 Coarse %')
ax0.legend()

# Anomaly with potential event thresholds marked
ax1.plot(GS_depth, anomaly_coarse)
ax1.axhline(0, color='k', lw = 0.7, label='zero')
ax1.axhline(np.median(anomaly_coarse) + np.std(anomaly_coarse), color='k', lw = 0.7, linestyle='dashed', label=r'1$\sigma$')
ax1.axhline(np.median(anomaly_coarse) + 2*np.std(anomaly_coarse), color='k', lw = 0.7, linestyle='dotted', label=r'2$\sigma$')
ax1.grid(color='lightgray', linestyle=':', lw=0.7, axis='both', zorder=0)
ax1.set_ylabel('Coarse Anomaly (%)')
ax1.set_xlabel('depth (cm)')
ax1.set_title('HOLY4 Coarse Anomaly')
ax1.legend()
plt.show()
```

# Part 3: Make Event Index

The problem with the above graph is that it counts every data point above the threshold as an event. Some events, though, can deposit several (or many!) cm of sand, meaning that consecutive 'events' as seen above may actually just be multiple data points that lie in the same event. So, we need to make a new event index and account for this.

```python
bool_anom = np.zeros(np.size(anomaly_coarse)) # make array that will populate with Boolean index

event_depth_array = [] # make empty depth array to save event depths in
event_gs_array = [] # make empty gs array to save event grainsizes in
event_anom_array = [] # make empty anomaly array
event_time_array = [] # make empty time array

# Each depth is either an event or not an event. This loop will find the events that exceed our 83rd percentile threshold.
# When there are several consecutive depths that exceed the threshold, the one with the highest raw GS will be coded as the event.

above_thresh = anomaly_coarse >= lowera
i = 0
n = len(anomaly_coarse)

while i < n:
    if above_thresh[i]:
        # Find the end of this contiguous run of depths above threshold
        j = i
        while j + 1 < n and above_thresh[j + 1]:
            j += 1
        
        # Pick the index within [i, j] with the greatest coarse_raw value
        peak_idx = i + np.argmax(coarse_raw[i:j+1])
        
        bool_anom[peak_idx] = 1
        event_depth_array.append(GS_depth[peak_idx])
        event_gs_array.append(coarse_raw[peak_idx])
        event_anom_array.append(anomaly_coarse[peak_idx])
        event_time_array.append(ages_med[peak_idx])
        
        i = j + 1
    else:
        i += 1
```

# Part 4: Gaussian Kernel Rate Estimation

The Gaussian Kernel Rate Estimation (GKRE) applies a sliding window to events over time to determine event frequency. The usual way of doing this, the Uniform Kernel, applies rigid boxes to each century, which creates super choppy frequency estimations. The way this works is that for each year, the Gaussian kernel looks at the years around it, adding up all of the events that happened within that range. It gives more weight to events that happened closer (in time) and less to those that happened further away. Once these weighted contributions are added, it scales it to 100 years.

```python
#%% Gaussian Kernel Rate Estimation
start_time = time.time()

def rate_parameter(allyrs, eventyrs, h):
    l = np.zeros(len(allyrs))
    for i in range(len(allyrs)):
        y = (allyrs[i] - eventyrs) / h
        l[i] = np.sum(np.power((2 * np.pi), -0.5) * np.exp((-y**2) / 2)) / h
    return l * 100 # storms per century

# We try a few different bandwidths (larger number means its a wider window)
l25 = rate_parameter(allyrs, eventyrs, 25)
l50 = rate_parameter(allyrs, eventyrs, 50)
l75 = rate_parameter(allyrs, eventyrs, 75)
l100 = rate_parameter(allyrs, eventyrs, 100)

print('time to run kernel rate estimation is ', (time.time() - start_time)/60, ' minutes')
```

## 4.1 Bandwidth Comparison

```python
#%% Bandwidth Sensitivity Comparison Plot (h = 25, 50, 75, 100 yr)

fig, ax = plt.subplots(figsize=(13, 5), dpi=150)

# Background storm event stems
ax.stem(allyrs, strmyrs, 'black', markerfmt=" ", basefmt=" ", label='storm occurrences')

# Overlay all three bandwidth curves
ax.plot(allyrs, l25, color='darksalmon', linewidth=1.3, alpha=0.85, label='h = 25 yr (sharp / multi-decadal')
ax.plot(allyrs, l50, color='darkseagreen', linewidth=1.3, label='h = 50 yr (centennial baseline)')
ax.plot(allyrs, l75, color='plum', linewidth=1.3, label='h = 75 yr (centennial smooth)')
ax.plot(allyrs, l100, color='lightsteelblue', linewidth=1.3, label='h = 100 yr (multi-centennial smooth)')

ax.set_xlabel('Years Before Present (YBP)')
ax.set_ylabel(r'Storm Frequency ($\lambda$, storms/century)')
ax.set_title('Bandwidth Sensitivity Comparison: h = 25, 50, and 100 Years (HOLY4)')
ax.grid(color='lightgray', linestyle=':', linewidth=0.7)
ax.set_xlim(min(allyrs) - 50, max(allyrs) + 50)
ax.legend(loc='upper left', ncol=1, fontsize=8, framealpha=0.92)
plt.tight_layout()
plt.savefig('HOLY4_Bandwidth_Comparison.png', dpi=300)
plt.show()
```

This plot is a comparison of the different sized windows that can be applied to the storm record and how smooth they each end up being. The orange is sharp and responsive to short term changes, the green is slightly smoother and responds slightly less to each individual event. The purple has a smoother profile and remains slightly responsive, and the blue is a very smooth curve.

## 4.3 Bootstrapping without Age Uncertainty

This portion will account for sampling uncertainty, but not age uncertainty. Since we're only looking at one core, we only have one dataset from one pond. In order to make sure that those results make sense we need to think about whether or not some hurricanes might have missed the exact spot or if extra ones hit right on top of it from pure luck. So we ask, 'is this increase in frequency real, or is this just from two lucky storms that happened to hit right there?'. To figure this out, we are going to sample 69 times from our pool of 69 events with replacement to test how much randomness will change our result. This will tell us if there are real stormy periods or just random noise. 

FYI: this cell does take a minute or two to run, so don't worry if nothing is happening right away. 

```python
# Bootstrap Without Age Uncertainty

start_time = time.time()

n_boot = 1000
h = 50
n_events = len(eventyrs)

# Matrix to store all 1,000 simulated storm-frequency curves (rows = iterations, cols = years)
boot_l50 = np.zeros((n_boot, len(allyrs)))

rng = np.random.default_rng(seed=42) # reproducible random draws

for b in range(n_boot):
    # Resample median event ages with replacement
    sampled_events = rng.choice(eventyrs, size=n_events, replace=True)
    boot_l50[b, :] = rate_parameter(allyrs, sampled_events, h)

# Calculate 5th and 95th percentiles (90% confidence envelope) at each year
ci_lower_90 = np.percentile(boot_l50, 5, axis=0)
ci_upper_90 = np.percentile(boot_l50, 95, axis=0)

# Optional: 16th and 84th percentiles (1-sigma / 68% confidence envelope)
ci_lower_68 = np.percentile(boot_l50, 16, axis=0)
ci_upper_68 = np.percentile(boot_l50, 84, axis=0)

print(f"Bootstrapping ({n_boot} iterations) completed in {(time.time() - start_time):.1f} seconds")
```

### 4.3.1 Plot GKRE with Bootstrapping

```python
fig, ax = plt.subplots(figsize=(10, 4.5), dpi=150)

# Stem plot and median GKRE curve
ax.stem(allyrs, strmyrs, 'k', markerfmt=" ", basefmt=" ")
ax.plot(allyrs, l50, color='black', linewidth=1.8, label='Storm frequency (h = 50 yr)', zorder=4)

# 90% and 68% bootstrap confidence bands
ax.fill_between(allyrs, ci_lower_90, ci_upper_90, color='gray', alpha=0.25, label='90% CI (no age uncertainty)', zorder=2)
ax.fill_between(allyrs, ci_lower_68, ci_upper_68, color='gray', alpha=0.35, label='68% CI (no age uncertainty)', zorder=3)

# Warm intervals (red)
ax.axvspan(xmin=5000, xmax=7500, color='tab:red', alpha=0.12, label='Warming', zorder=1)
ax.axvspan(xmin=1600, xmax=2200, color='tab:red', alpha=0.12, zorder=1)
ax.axvspan(xmin=700, xmax=1000, color='tab:red', alpha=0.12, zorder=1)

# Cool intervals (blue)
ax.axvspan(xmin=3900, xmax=4200, color='tab:blue', alpha=0.14, label='Cooling', zorder=1)
ax.axvspan(xmin=100, xmax=650, color='tab:blue', alpha=0.14, zorder=1)

# Subtle labels along the top of each band
ymax = np.max(ci_upper_90) * 1.12
ax.set_ylim(0, ymax)
label_y = ymax * 0.93

intervals = [
    (6250, 'HTM', 'darkred'),
    (4050, '4.2 ka', 'darkblue'),
    (1900, 'RWP', 'darkred'),
    (850,  'MCA', 'darkred'),
    (375,  'LIA', 'darkblue')
]
for x_pos, text, col in intervals:
    ax.text(x_pos, label_y, text, ha='center', va='top', fontsize=8, fontweight='bold', color=col)

ax.set_xlabel('Years Before Present (YBP)')
ax.set_ylabel(r'Storm Frequency ($\lambda$, storms/century)')
ax.grid(color='lightgray', linestyle=':', linewidth=0.7, axis='both', which='major', zorder=0)
ax.set_xlim(min(allyrs) - 50, max(allyrs) + 50)
ax.legend(
    bbox_to_anchor=(10, 5.5),
    bbox_transform=ax.transData,
    loc='center left'
)
plt.tight_layout()
plt.show()
```

## 4.4 Bootstrapping with Age Uncertainty

So now, we have accounted for sampling uncertainty, but we need to account for chronological uncertainty. Age modeling has error, and without accounting for that, we are assuming that each of the storms indeed happened on the median age reported by the age model. The age model gives us  What we do now is repeat the resampling, but this time, we randomly perturb the age of each event slightly within its error.

```python
# Bootstrap With Age Uncertainty (Bacon / rPlum MCMC + Resampling)

start_time = time.time()

# Load raw MCMC age distributions exported from R: shape = (n_mcmc, n_events)
age_dist_raw = np.genfromtxt("HOLY4_Age_Distributions.csv", delimiter=",", skip_header=1)
age_dist = age_dist_raw[:, 1:]  # drop leading R row-index column

print("Loaded age_dist shape (MCMC iterations x events):", age_dist.shape)
assert age_dist.shape[1] == len(eventyrs), (
    f"Column count ({age_dist.shape[1]}) does not match number of events ({len(eventyrs)})!"
)

n_mcmc, n_events = age_dist.shape
n_boot = 1000
h = 50

# Matrix to store all 1,000 age-uncertain storm-frequency curves
boot_age_l50 = np.zeros((n_boot, len(allyrs)))

rng = np.random.default_rng(seed=42)

for b in range(n_boot):
    # 1. Draw a random MCMC age realization for each event (accounting for age-depth uncertainty)
    rand_mcmc_rows = rng.integers(0, n_mcmc, size=n_events)
    mcmc_event_ages = age_dist[rand_mcmc_rows, np.arange(n_events)]
    
    # 2. Bootstrap resample those age-perturbed events with replacement
    sampled_events = rng.choice(mcmc_event_ages, size=n_events, replace=True)
    
    # 3. Compute GKRE storm frequency curve
    boot_age_l50[b, :] = rate_parameter(allyrs, sampled_events, h)

# Calculate 90% (5th–95th) and 68% (16th–84th) confidence envelopes
age_ci_lower_90 = np.percentile(boot_age_l50, 5, axis=0)
age_ci_upper_90 = np.percentile(boot_age_l50, 95, axis=0)

age_ci_lower_68 = np.percentile(boot_age_l50, 16, axis=0)
age_ci_upper_68 = np.percentile(boot_age_l50, 84, axis=0)

# Optional: save final GKRE + confidence bounds to CSV for downstream plotting/tables
results_df = pd.DataFrame({
    "Year_YBP": allyrs,
    "Storm_Events": strmyrs,
    "Lambda_h25": l25,
    "Lambda_h50": l50,
    "Lambda_h100": l100,
    "NoAge_CI68_Low": ci_lower_68,
    "NoAge_CI68_High": ci_upper_68,
    "NoAge_CI90_Low": ci_lower_90,
    "NoAge_CI90_High": ci_upper_90,
    "AgeUnc_CI68_Low": age_ci_lower_68,
    "AgeUnc_CI68_High": age_ci_upper_68,
    "AgeUnc_CI90_Low": age_ci_lower_90,
    "AgeUnc_CI90_High": age_ci_upper_90,
})
results_df.to_csv("HOLY4_Storm_Frequency_Results.csv", index=False)

print(f"Age-uncertain bootstrapping completed in {(time.time() - start_time):.1f} seconds")
```

### 4.4.1 Plot Storm Frequency with Sampling & Chronological Uncertainty

```python
fig, ax = plt.subplots(figsize=(10.5, 4.8), dpi=150)

# Stem plot and median GKRE curve
ax.stem(allyrs, strmyrs, 'k', markerfmt=" ", basefmt=" ")
ax.plot(allyrs, l50, color='black', linewidth=1.8, label='Median age storm frequency (h = 50 yr)', zorder=4)

# 90% and 68% age-uncertain bootstrap confidence bands
ax.fill_between(allyrs, age_ci_lower_90, age_ci_upper_90, color='dimgray', alpha=0.22, label='90% CI (with age uncertainty)', zorder=2)
ax.fill_between(allyrs, age_ci_lower_68, age_ci_upper_68, color='dimgray', alpha=0.35, label='68% CI (with age uncertainty)', zorder=3)

# Warm intervals (red)
ax.axvspan(xmin=5000, xmax=7500, color='tab:red', alpha=0.12, label='Warming', zorder=1)
ax.axvspan(xmin=1600, xmax=2200, color='tab:red', alpha=0.12, zorder=1)
ax.axvspan(xmin=700, xmax=1000, color='tab:red', alpha=0.12, zorder=1)

# Cool intervals (blue)
ax.axvspan(xmin=3900, xmax=4200, color='tab:blue', alpha=0.14, label='Cooling', zorder=1)
ax.axvspan(xmin=100, xmax=650, color='tab:blue', alpha=0.14, zorder=1)

# Subtle labels along the top of each band
ymax = np.max(age_ci_upper_90) * 1.14
ax.set_ylim(0, ymax)
label_y = ymax * 0.94

intervals = [
    (6250, 'HTM', 'darkred'),
    (4050, '4.2 ka', 'darkblue'),
    (1900, 'RWP', 'darkred'),
    (850,  'MCA', 'darkred'),
    (375,  'LIA', 'darkblue')
]
for x_pos, text, col in intervals:
    ax.text(x_pos, label_y, text, ha='center', va='top', fontsize=8, fontweight='bold', color=col)

ax.set_xlabel('Years Before Present (YBP)')
ax.set_ylabel(r'Storm Frequency ($\lambda$, storms/century)')
ax.grid(color='lightgray', linestyle=':', linewidth=0.7, axis='both', which='major', zorder=0)
ax.set_xlim(min(allyrs) - 50, max(allyrs) + 50)
ax.legend(
    bbox_to_anchor=(10, 4.5),
    bbox_transform=ax.transData,
    loc='center left'
)
plt.tight_layout()
plt.savefig("HOLY4_Storm_Frequency_Final.png", dpi=300)
plt.show()
```

#### 4.4.2 Summary Table

```python
import pandas as pd
import numpy as np

res = pd.read_csv("HOLY4_Storm_Frequency_Results.csv")

periods = [
    ("Full Record (All Years)", res["Year_YBP"].min(), res["Year_YBP"].max()),
    ("HTM (Warm: 7500–5000 YBP)", 5000, 7500),
    ("4.2 ka (Cool: 4200–3900 YBP)", 3900, 4200),
    ("RWP (Warm: 2200–1600 YBP)", 1600, 2200),
    ("MCA (Warm: 1000–700 YBP)", 700, 1000),
    ("LIA (Cool: 650–100 YBP)", 100, 650),
]

summary = []
for name, start, end in periods:
    sub = res[(res["Year_YBP"] >= start) & (res["Year_YBP"] <= end)]
    summary.append({
        "Climate Period": name,
        "Duration (yr)": len(sub),
        "Event Count": int(sub["Storm_Events"].sum()),
        "Mean Rate (l50)": round(sub["Lambda_h50"].mean(), 2),
        "Peak Rate (l50)": round(sub["Lambda_h50"].max(), 2),
        "Mean 68% CI": f"{sub['AgeUnc_CI68_Low'].mean():.2f}–{sub['AgeUnc_CI68_High'].mean():.2f}",
        "Mean 90% CI": f"{sub['AgeUnc_CI90_Low'].mean():.2f}–{sub['AgeUnc_CI90_High'].mean():.2f}",
    })

summary_df = pd.DataFrame(summary)
print(summary_df.to_string(index=False))
summary_df.to_csv("HOLY4_Climate_Interval_Summary.csv", index=False)
```
