```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

## Data Description and Access

The data I am using for my project is my primary data. It consists of data from a 7-meter long core and the accompanying 1-meter surface drive, which have been used to form a composite complete core, HOLY4. The collected data includes grain size (% > 125um), organic matter (% lost on ignition), and median age (age control performed by sampling for 14C and 210Pb and using the rPlum and rBACON softwares). 

This core, HOLY4, along with 5 others, was collected from Holyrood Pond, a back-barrier pond in Newfoundland, Atlantic Canada. Subsequent labwork was performed in the Wallace Lab by Kaite Gross from May 2024 to August 2024. Sampling was done in 1cm intervals for the length of the core.

```python
infile = '/Users/ariellenogueira/Desktop/Dissertation/grain size/raw data/Data_HOLY4_cleaned_093026.csv'

data = pd.read_csv(infile, sep=',', low_memory = False)
```

### Visualizing Data

```python
fig, axes = plt.subplots(2, 1,figsize=(28,10),layout="constrained")
sns.lineplot(
    data=data,
    x = "CorrectedDepth",
    y = "Organic%",
    ax=axes[0],
)
axes[0].set_title(
    "Organic Matter"
)
sns.lineplot(
    data=data,
    x = "CorrectedDepth",
    y = "Coarse%",
    ax=axes[1],
)
axes[1].set_title(
    "% >125um"
)
axes[0].set_xlabel('Depth (cm)')
axes[0].set_ylabel('Organic Matter Content (%)')
axes[1].set_xlabel('Depth (cm)')
axes[1].set_ylabel('Organic Matter Content (%)')
```

#### Figure 1

Plot of organic matter and coarse percent, by depth. Organic matter content is significantly low throughout the core, with some spikes due to clusters of roots and leaves. Coarse percent is highly variable, with noticable peaks representing potential storm surge deposits, which appear as coarse peaks against a finer background sediment. 

## Spatial and Temporal Coverage

These data cover thousands of years of potential hurricane history for the North Atlantic, specifically from Holyrood Pond in Newfoundland. They represent only a small subset of this history, and part of the goal of the larger project (taking uncertainty into account) is to examine the statistical significance of this small snapshot, as well as use a novel form of data analysis to standardize these data to compare to other studies done in the region. 

```python
fig, axes = plt.subplots(1,layout="constrained")
sns.lineplot(
    data=data,
    y = "CorrectedDepth",
    x = "C14MedAge",
)
plt.title("Age Depth Plot")
axes.set_xlabel('Age (yr BP)')
axes.set_ylabel('Depth (cm)')
```

#### Figure 2

Age-depth plot showing that the 850cm core covers up to 7500 yr BP. There is a relatively consistent sedimentation rate until 4000 BP, when it decreases.
